## Problem 1 — Array creation and shapes

## Problem 2 — Vectorized evaluation of a function
Let $f(x) = x^3 - 3x + 2$. Create a vector `x` over [-3, 3] step size 0.1, with both endpoints included. 
Then compute `f(x)` as a vector, without loops.

```python

# TODO: (1) Define the function f(x). 
# (2) Build x with np.linspace(...) and print the result. 
# (3)  Compute f(x) as a vector and print the result.

def f(x):
    return x**3 - 3*x + 2

x = np.linspace(-3, 3, 61)   # 61 points: -3.0, -2.9, ..., 3.0
print(x)

fx = f(x)
print(fx)
```
