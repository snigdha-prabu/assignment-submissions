## Problem 1 — Read and summarize
Open `sample_numbers.txt`, read integers line-by-line, and compute:
- count, min, max, mean

Print the results.

```python

# TODO: Implement the file read & summary.
# Reminders (see Mini-exercise discussed in Files and Modules video):
# with open("sample_numbers.txt", "r", encoding="utf-8") as f:
#     for line in f:
#         ...
count = 0
total = 0
min_val = None
max_val = None

with open("sample_numbers.txt", "r", encoding="utf-8") as f:
    for line in f:
        num = int(line.strip())
        count += 1
        total += num
        if min_val is None or num < min_val:
            min_val = num
        if max_val is None or num > max_val:
            max_val = num

mean = total / count

print(f"count: {count}")
print(f"min: {min_val}")
print(f"max: {max_val}")
print(f"mean: {mean}")

```

Write a fictional report `summary.txt` that contains at least two lines of text.

Then append it to add to add a conclusion.

Then read the file and print its contents to verify that your code behaves as expected.


```python
# Write summary.txt using mode "w" (overwrite)
with open("summary.txt", "w", encoding="utf-8") as f:
    f.write("This is a fictional report.\n")
    f.write("It contains at least two lines of text.\n")

# Append a conclusion line
with open("summary.txt", "a", encoding="utf-8") as f:
    f.write("End of report.\n")

# Read the file and print its contents to verify
with open("summary.txt", "r", encoding="utf-8") as f:
    contents = f.read()
    print(contents)
```

## Problem 3 — A tiny module
Create a file `mymathutils.py` with a function that computes the mean of a list if it is nonempty, and returns `None` otherwise.

```python
# Create the module file programmatically, then import and use it.
module_code = '''def safe_mean(l):
    if len(l) == 0:
        return None
    return sum(l) / len(l)'''

with open("mymathutils.py", "w", encoding="utf-8") as f:
    f.write(module_code)

# Now import and test
from mymathutils import safe_mean
print(safe_mean([1, 3, 5, 9]))
print(safe_mean([]))
```
